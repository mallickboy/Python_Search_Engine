```python
import requests

def pipeline(query):

    # s1 - get vector
    response = requests.get(f"http://192.168.1.3:4002/vectorizer?sentence={query}")
    response = response.json()  
    vector= response.get("vector",[])
    # print(len(vector))
    print(vector)

    # s2 - send vector to search
    response = requests.post("http://192.168.1.3:4001/search", json={"query_vector": vector}).json()
    print(response[0].get("h1"), response[0].get("h2"))
    print(len(response))

pipeline("dictionary")
```
